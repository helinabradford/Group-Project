## 7. Filter the Data

```python
clean = df.drop_duplicates().copy()

# Standardize store names: strip spaces, lowercase, then map variants to 3 stores
clean["store"] = (clean["store"].str.strip().str.lower()
                  .replace({"campus store": "campus", "down town": "downtown", "airport t2": "airport"})
                  .str.title())

# Convert timestamp to datetime
clean["timestamp"] = pd.to_datetime(clean["timestamp"])

# Give the meaningful 'missing' values a clear label
clean["size"] = clean["size"].fillna("No size")
clean["discount_code"] = clean["discount_code"].fillna("None")

print("Rows before:", len(df), "| after removing duplicates:", len(clean))
display(clean["store"].value_counts())
print(clean["timestamp"].dtype)
```

```python
# line_total = unit_price * quantity (revenue for each row, before any discount)
clean["line_total"] = (clean["unit_price"] * clean["quantity"]).round(2)

# Also pull the hour and month out of the timestamp
clean["hour"] = clean["timestamp"].dt.hour
clean["month"] = clean["timestamp"].dt.month_name()

display(clean[["transaction_id", "unit_price", "quantity", "line_total", "hour", "month"]].head())
```

```python
# Filter 1: returns / refunds (negative quantity)
returns = clean[clean["quantity"] < 0]
print("Return rows:", len(returns), "| money returned:", round(returns["line_total"].sum(), 2))
display(returns.head())
```

```python
# Filter 2: Campus transactions that used the STUDENT10 discount code
campus_student = clean[(clean["store"] == "Campus") & (clean["discount_code"] == "STUDENT10")]
print("Campus STUDENT10 rows:", len(campus_student))
display(campus_student.head())
```

```python
# Filter 3: zero-price rows
zero_price = clean[clean["unit_price"] == 0]
print("Zero-price rows:", len(zero_price))
display(zero_price.head())
```

```python
# Filter 4: regular sales only (positive quantity and price), used for the analysis below
sales = clean[(clean["quantity"] > 0) & (clean["unit_price"] > 0)]
print("Rows kept as regular sales:", len(sales), "of", len(clean))
```

```python
# **Subsets and why they are interesting:**
# - **Returns** (negative quantity) isolate the refunds so I can see how often they happen and how much they affect revenue.
# - **Campus STUDENT10** looks at the student discount users at the campus store, a group the business may want to understand or target.
# - **Zero-price rows** isolate the suspicious rows so I can check whether they are comped items or errors.
# - **Regular sales** removes the returns and zero-price rows so my averages and totals reflect real purchases.
```

## 9. Grouping and Aggregation

```python
df['revenue'] = df['unit_price'] * df['quantity']
```

```python
# Group the data by product category and calculate summary statistics
category_summary = df.groupby('category').agg(
    total_transactions=('transaction_id', 'count'),
    total_quantity_sold=('quantity', 'sum'),
    total_revenue=('revenue', 'sum'),
    average_unit_price=('unit_price', 'mean')
).round(2)
```

```python
print(category_summary)
```

On Campus had the most transactions, while the Airport had the highest average per sale. So, airport customers are willing to spend more.
